```python
# 1. ตั้งค่าฟอนต์ภาษาไทย (สำหรับ Windows ใช้ Tahoma, Mac ใช้ 'Ayuthaya' หรืออื่นๆ ที่รองรับ)
plt.rcParams['font.family'] = 'Tahoma'

# 2. นำเข้าและทำความสะอาดข้อมูล (Data Cleaning)
df = pd.read_excel('bangkok_traffic_2560_Present.xlsx')

# ตัดแถวแรกที่เป็น header ขยะออก และตั้งชื่อคอลัมน์ใหม่ให้ทำงานง่ายขึ้น (ภาษาอังกฤษ)
df_clean = df.iloc[1:].copy()
df_clean.columns = ['ID', 'Survey_date', 'Date', 'Month', 'Year', 'Covid_period', 'Intersection', 'Road', 'period','time_period',
                    'passenger_car', 'pickup_van', 'large_bus', 'Small_Bus', 'Truck', 'tuk_tuk']

# แปลงชนิดข้อมูลให้เป็นตัวเลขสำหรับการคำนวณ
numeric_cols = ['passenger_car', 'pickup_van', 'large_bus', 'Small_Bus', 'Truck', 'tuk_tuk']
for col in numeric_cols:
    df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')

# ลบแถวที่ไม่มีข้อมูลปี (Missing values handling)
df_clean = df_clean.dropna(subset=['Year'])

# 3. EDA วิเคราะห์และสร้างกราฟ

# --- รูปที่ 1: แนวโน้มปริมาณรถแต่ละประเภทแยกตามปี ---
veh_cols = ['passenger_car', 'pickup_van', 'large_bus', 'Small_Bus', 'Truck', 'tuk_tuk']
yearly_trend = df_clean.groupby('Year')[veh_cols].sum()

plt.figure(figsize=(10, 6))
yearly_trend.plot(kind='bar', stacked=False, colormap='tab10', ax=plt.gca())
plt.title('Figure 1: Total Traffic Volume by Year and Vehicle Type (2022 - 2026)', fontsize=14)
plt.xlabel('ปี ค.ศ.', fontsize=12)
plt.ylabel('ปริมาณรถรวม (คัน)', fontsize=12)
plt.legend(title='ประเภทรถ', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

# --- รูปที่ 2: ปริมาณรถแยกตามช่วงเวลา ---
time_trend = df_clean.groupby('Time_period')[veh_cols].sum()

plt.figure(figsize=(10, 6))
time_trend.plot(kind='bar', stacked=false, colormap='Set2', ax=plt.gca())
plt.title('Figure 2: Traffic Volume Distribution by Time of Day', fontsize=14)
plt.xlabel('ช่วงเวลา', fontsize=12)
plt.ylabel('ปริมาณรถรวม (คัน)', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.legend(title='ประเภทรถ', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

# --- รูปที่ 3: Top 10 ทางแยกที่มีปริมาณรถหนาแน่นที่สุด ---
top_intersections = df_clean.groupby('Intersection')['Vol_by_time'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12, 6))
import seaborn as sns
sns.barplot(x=top_intersections.values, y=top_intersections.index, palette='viridis')
plt.title('Figure 3: Top 10 Intersections with Highest Traffic Volume', fontsize=14)
plt.xlabel('ปริมาณรถรวม (คัน)', fontsize=12)
plt.ylabel('บริเวณทางแยก', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import seaborn as sns

# 1. ติดตั้งและโหลดฟอนต์ภาษาไทยสำหรับ Google Colab / Linux
!apt-get -y install fonts-thai-tlwg > /dev/null 2>&1
font_path = '/usr/share/fonts/truetype/tlwg/Garuda.ttf'
fm.fontManager.addfont(font_path)

# 2. ตั้งค่าธีมการแสดงผล
sns.set_theme(style='whitegrid', font='Garuda')
plt.rcParams['axes.unicode_minus'] = False

# 3. นำเข้าข้อมูลและเตรียมคอลัมน์
excel_path = 'bangkok_traffic_2560_Present.xlsx'
df = pd.read_excel(excel_path, sheet_name='traffic_data')

veh_cols = ['passenger_car', 'pickup_van', 'large_bus', 'small_bus', 'truck', 'tuk_tuk']
veh_labels = {
    'passenger_car': 'รถยนต์นั่ง',
    'pickup_van': 'รถกระบะ/ตู้',
    'large_bus': 'รถบัสใหญ่',
    'small_bus': 'รถบัสเล็ก',
    'truck': 'รถบรรทุก',
    'tuk_tuk': 'รถตุ๊กตุ๊ก'
}

for col in veh_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0)

# ทำความสะอาดข้อมูล Outlier
df_clean = df[df['pickup_van'] < 1000000].copy()
df_clean['Vol_by_time'] = df_clean[veh_cols].sum(axis=1)

# 4. จัด Layout แบบ Asymmetric ด้วย subplot_mosaic (เน้นกราฟหลักด้านซ้ายให้ใหญ่กว่า)
fig = plt.figure(figsize=(16, 9), dpi=120)
ax_dict = fig.subplot_mosaic(
    [
        ["main", "top_right"],
        ["main", "bottom_right"]
    ],
    gridspec_kw={'width_ratios': [1.4, 1], 'wspace': 0.25, 'hspace': 0.35}
)

fig.suptitle('Bangkok Traffic Analytical Dashboard (2560 - Present)', fontsize=18, fontweight='bold', y=0.96)

# ==============================================================================
# 1. Main Axis (กราฟหลัก - ขนาดใหญ่กว่า): Top 10 ทางแยกที่มีปริมาณรถหนาแน่นที่สุด
# ==============================================================================
ax_main = ax_dict['main']
top10 = df_clean.groupby('intersection')['Vol_by_time'].sum().sort_values(ascending=False).head(10)

sns.barplot(x=top10.values / 1e3, y=top10.index, ax=ax_main, palette='mako')
ax_main.set_title('★ [Main Focus] Top 10 ทางแยกที่มีปริมาณรถหนาแน่นที่สุด', fontsize=14, fontweight='bold', pad=12)
ax_main.set_xlabel('ปริมาณรถสะสมรวม (พันคัน)', fontsize=11)
ax_main.set_ylabel('')
sns.despine(ax=ax_main, top=True, right=True)

# ใส่ตัวเลขแสดงค่าปลายแท่งกราฟ
for p in ax_main.patches:
    width = p.get_width()
    ax_main.annotate(f'{width:,.0f}k คัน',
                     (width, p.get_y() + p.get_height() / 2.),
                     ha='left', va='center', xytext=(6, 0),
                     textcoords='offset points', fontsize=10, fontweight='bold')

# ==============================================================================
# 2. Top Right Axis: แนวโน้มการใช้งานรถแต่ละประเภทรายปี (Line Chart)
# ==============================================================================
ax_top = ax_dict['top_right']
yearly = df_clean.groupby('year')[veh_cols].sum() / 1e3
yearly.columns = [veh_labels[col] for col in yearly.columns]

yearly.plot(kind='line', marker='o', linewidth=2, ax=ax_top, colormap='tab10')
ax_top.set_title('แนวโน้มการใช้งานยานพาหนะจำแนกตามรายปี', fontsize=12, fontweight='bold', pad=10)
ax_top.set_xlabel('ปี ค.ศ.', fontsize=10)
ax_top.set_ylabel('ปริมาณรถรวม (พันคัน)', fontsize=10)
ax_top.legend(title='ประเภทรถ', frameon=True, facecolor='white', fontsize=8, loc='upper right')
sns.despine(ax=ax_top, top=True, right=True)

# ==============================================================================
# 3. Bottom Right Axis: สัดส่วนปริมาณรถแยกตามช่วงเวลา (Grouped Bar Chart)
# ==============================================================================
ax_bottom = ax_dict['bottom_right']
period_data = df_clean.groupby('period')[veh_cols].sum() / 1e3
period_data.columns = [veh_labels[col] for col in period_data.columns]

period_data.plot(kind='bar', ax=ax_bottom, colormap='Set2', width=0.75)
ax_bottom.set_title('ปริมาณรถแยกตามช่วงเวลา (Period)', fontsize=12, fontweight='bold', pad=10)
ax_bottom.set_xlabel('ช่วงเวลา', fontsize=10)
ax_bottom.set_ylabel('ปริมาณรถรวม (พันคัน)', fontsize=10)
ax_bottom.set_xticklabels(ax_bottom.get_xticklabels(), rotation=0)
ax_bottom.legend(title='ประเภทรถ', frameon=True, facecolor='white', fontsize=8)
sns.despine(ax=ax_bottom, top=True, right=True)

plt.show()

# 4. ดึงข้อมูล Top 10 ทางแยก และจัดทำ Pivot Table Matrix
top10_intersections = df_clean.groupby('intersection')['Vol_by_time'].sum().sort_values(ascending=False).head(10).index
df_top10 = df_clean[df_clean['intersection'].isin(top10_intersections)]

# Pivot Table: ความหนาแน่นของรถ (ทางแยก x ช่วงเวลา) ในหน่วยพันคัน
pivot_heatmap = df_top10.pivot_table(index='intersection', columns='period', values='Vol_by_time', aggfunc='sum') / 1e3
pivot_heatmap = pivot_heatmap[['เร่งด่วนเช้า', 'นอกเร่งด่วน', 'เร่งด่วนเย็น']] # เรียงลำดับช่วงเวลา

# 5. สร้างกราฟ Heatmap เดี่ยว (Figure 4)
plt.figure(figsize=(11, 7), dpi=120)

sns.heatmap(
    pivot_heatmap,
    annot=True,
    fmt=',.1f',
    cmap='YlOrRd',
    linewidths=1.5,
    linecolor='white',
    cbar_kws={'label': 'ปริมาณรถสะสมรวม (พันคัน)'},
    annot_kws={'size': 10, 'weight': 'bold'}
)

plt.title('Figure 4: Traffic Volume Heatmap Across Top 10 Intersections & Periods', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('ช่วงเวลา (Period)', fontsize=11, labelpad=10)
plt.ylabel('บริเวณทางแยก (Intersection)', fontsize=11, labelpad=10)
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()

# ทำความสะอาดและหาผลรวมตามทางแยก
df_clean = df[df['pickup_van'] < 1000000].copy()
df_clean['Vol_by_time'] = df_clean[veh_cols].sum(axis=1)

summary = df_clean.groupby('intersection').agg({
    'passenger_car': 'sum',
    'truck': 'sum',
    'large_bus': 'sum',
    'pickup_van': 'sum',
    'Vol_by_time': 'sum'
}).reset_index()

summary['commercial'] = summary['truck'] + summary['large_bus']
summary['pickup_pct'] = (summary['pickup_van'] / summary['Vol_by_time']) * 100

# เลือกทางแยกที่มีปริมาณรถสูงสุด 20 อันดับแรก
top20 = summary.sort_values(by='Vol_by_time', ascending=False).head(20)

# 2. พล็อต Bubble Plot 4 มิติ
plt.figure(figsize=(10, 6))

x = top20['passenger_car'] / 1000       # แกน X: รถส่วนบุคคล (พันคัน)
y = top20['commercial'] / 1000          # แกน Y: รถเชิงพาณิชย์ (พันคัน)
sizes = (top20['Vol_by_time'] / 1000) * 2.5  # ขนาดวงกลม: ปริมาณรถรวม
colors = top20['pickup_pct']            # สีวงกลม: % รถกระบะ/ตู้

scatter = plt.scatter(x, y, s=sizes, c=colors, cmap='viridis', alpha=0.6, edgecolors='black')

cbar = plt.colorbar(scatter)
cbar.set_label('Pickup Truck Ratio (%)')

plt.title('Bubble Plot: 4-Variable Traffic Analysis by Intersection')
plt.xlabel('Passenger Cars Volume (Thousands)')
plt.ylabel('Commercial Vehicles Volume (Thousands)')
plt.grid(True, linestyle='--', alpha=0.5)

# 2. แปลงข้อมูลเป็น Long Format
df_melted = df_clean.melt(
    id_vars=['intersection'],
    value_vars=veh_cols,
    var_name='Vehicle_Type',
    value_name='Volume',
)

# ตั้งค่าคุณสมบัติของจุด Outlier (flierprops)
flierprops = dict(
    marker='o',
    markersize=2.5,  # ย่อขนาดจุดให้เล็กลง
    alpha=0.15,  # กำหนดความโปร่งใสที่ 15% (จุดทับซ้อนกันจะเข้มขึ้น)
    markeredgecolor='none',  # เอาเส้นขอบออกเพื่อไม่ให้บังกันเอง
    markerfacecolor='black',
)

plt.figure(figsize=(10, 6))
sns.boxplot(
    x='Vehicle_Type',
    y='Volume',
    data=df_melted,
    palette='Set2',
    flierprops=flierprops,  # ส่งค่าพารามิเตอร์เข้าไปใน boxplot
)

plt.yscale('log')
plt.title('Box Plot: Traffic Volume Distribution (Translucent Outliers)')
plt.xlabel('Vehicle Type')
plt.ylabel('Volume per Record (Log Scale)')
plt.grid(True, axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()

# จัดกลุ่มตามช่วงสถานการณ์
def get_era(year):
  if year < 2020:
    return '1. Pre-COVID (2017-2019)'
  elif year in [2020, 2021]:
    return '2. COVID Lockdown (2020-2021)'
  else:
    return '3. Post-COVID (2022+)'


df_target['Era'] = df_target['year'].apply(get_era)

# พล็อต กราฟแท่งแบบจัดกลุ่ม Era
plt.figure(figsize=(10, 5))
sns.barplot(
    data=df_target,
    x='Month_Label',
    y='Total_Volume',
    hue='Era',
    palette='Set2',
    errorbar=None,
)
plt.title(
    'Traffic Volume Comparison by COVID Era', fontsize=13, fontweight='bold'
)
plt.xlabel('เดือน')
plt.ylabel('ปริมาณรถเฉลี่ย (คัน)')
plt.legend(title='ช่วงเวลา', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.grid(True, axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()


plt.figure(figsize=(10, 5))
sns.lineplot(
    data=df_target,
    x='year',
    y='Total_Volume',
    hue='Month_Label',
    marker='o',
    linewidth=2.5,
    errorbar=None,
)
plt.title(
    'Yearly Traffic Volume Trend by Month', fontsize=13, fontweight='bold'
)
plt.xlabel('ปี (Year)')
plt.ylabel('ปริมาณรถเฉลี่ย (คัน)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(title='เดือน', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()

# เตรียม Pivot Table
pivot_df = df_target.groupby(['year', 'Month_Label'])[
    'Total_Volume'
].mean().unstack(level='Month_Label')
pivot_df = pivot_df[
    ['เม.ย. (Apr)', 'พ.ค. (May)', 'ก.ค. (Jul)', 'ส.ค. (Aug)', 'ก.ย. (Sep)']
]

# พล็อต Heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(
    pivot_df,
    annot=True,
    fmt='.0f',
    cmap='YlGnBu',
    cbar_kws={'label': 'ปริมาณรถเฉลี่ย (คัน)'},
    linewidths=0.5,
)
plt.title(
    'Traffic Volume Heatmap by Year & Month', fontsize=13, fontweight='bold'
)
plt.xlabel('เดือน')
plt.ylabel('ปี (Year)')
plt.tight_layout()
plt.show()
```
