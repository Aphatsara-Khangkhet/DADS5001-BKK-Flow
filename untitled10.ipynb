```python
# ==============================================================================
# 1. ติดตั้งฟอนต์ภาษาไทยบน Google Colab และดาวน์โหลดฟอนต์ Sarabun
# ==============================================================================
!apt-get install -y fonts-thai-tlwg > /dev/null 2>&1
!wget -q https://github.com/google/fonts/raw/main/ofl/sarabun/Sarabun-Regular.ttf -O Sarabun-Regular.ttf

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

# เพิ่มและตั้งค่าฟอนต์ภาษาไทยสำหรับ Matplotlib
fm.fontManager.addfont('Sarabun-Regular.ttf')
plt.rc('font', family='Sarabun')
plt.rcParams['font.size'] = 11

# ==============================================================================
# 2. อ่านไฟล์และประมวลผลข้อมูล (รวมรถทุกประเภท)
# ==============================================================================
file_path = 'bangkok_traffic_2560_Present (NotFinal).xlsx'
df = pd.read_excel(file_path)

# รวมคอลัมน์ปริมาณรถทุกประเภทเข้าด้วยกัน
vehicle_cols = ['passenger_car', 'pickup_van', 'large_bus', 'small_bus', 'truck', 'tuk_tuk']
df['total_vehicles'] = df[vehicle_cols].sum(axis=1)

# จัดกลุ่มรายปี คำนวณ Mean, SD, Count และ SEM (Standard Error of the Mean)
stats = df.groupby('year')['total_vehicles'].agg(['mean', 'std', 'count']).reset_index()

# คำนวณ SEM เพื่อให้สเกลความกว้างของแถบความคลาดเคลื่อนแคบและสมส่วน
stats['sem'] = stats['std'] / np.sqrt(stats['count'])

# ==============================================================================
# 3. สร้างกราฟเส้น (Line Graph) แสดงภาษาไทย
# ==============================================================================
fig, ax = plt.subplots(figsize=(10, 5))

# เส้นค่าเฉลี่ยปริมาณรถรวม
ax.plot(stats['year'], stats['mean'],
        marker='o', linewidth=2, color='#1f77b4', label='ปริมาณรถรวมเฉลี่ย (Mean)')

# แถบเงาความคลาดเคลื่อนมาตรฐาน (±1 SEM)
ax.fill_between(stats['year'],
                stats['mean'] - stats['sem'],
                stats['mean'] + stats['sem'],
                color='#1f77b4', alpha=0.25, label='ความคลาดเคลื่อนมาตรฐาน (±1 SEM)')

# กำหนดชื่อหัวข้อและแกนภาษาไทย
ax.set_title('แนวโน้มปริมาณรถทุกประเภทรวมกันเฉลี่ยรายปี (2560 - ปัจจุบัน)', fontsize=14, fontweight='bold', pad=12)
ax.set_xlabel('ปี (ค.ศ.)', fontsize=12)
ax.set_ylabel('จำนวนรถรวมทุกประเภท (คัน)', fontsize=12)
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(fontsize=11)

plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# โค้ดแสดง 5 อันดับทางแยกที่มีปริมาณจราจรสูงสุดและต่ำที่สุด
# ==============================================================================

# 1. ติดตั้งฟอนต์ภาษาไทยสำหรับ Google Colab
!apt-get install -y fonts-thai-tlwg > /dev/null 2>&1
!wget -q https://github.com/google/fonts/raw/main/ofl/sarabun/Sarabun-Regular.ttf -O Sarabun-Regular.ttf

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

# ตั้งค่าฟอนต์ Sarabun ให้แสดงภาษาไทยได้อย่างถูกต้อง
fm.fontManager.addfont('Sarabun-Regular.ttf')
plt.rc('font', family='Sarabun')
plt.rcParams['font.size'] = 11

# 2. อ่านไฟล์ข้อมูลและประมวลผล
file_path = 'bangkok_traffic_2560_Present (NotFinal).xlsx'
df = pd.read_excel(file_path)

# รวมคอลัมน์รถทุกประเภท
vehicle_cols = ['passenger_car', 'pickup_van', 'large_bus', 'small_bus', 'truck', 'tuk_tuk']
df['total_vehicles'] = df[vehicle_cols].sum(axis=1)

# คำนวณค่าเฉลี่ยปริมาณรถต่อบันทึกของแต่ละทางแยก
intersection_avg = df.groupby('intersection')['total_vehicles'].mean().reset_index()

# คัดเลือก Top 5 สูงสุด และ Top 5 ต่ำที่สุด
top5_high = intersection_avg.sort_values(by='total_vehicles', ascending=False).head(5)
top5_low = intersection_avg.sort_values(by='total_vehicles', ascending=True).head(5)

# รวมข้อมูลทั้งสองกลุ่มเพื่อนำมาพล็อตลงในกราฟเดียว
combined_df = pd.concat([top5_low, top5_high]).sort_values(by='total_vehicles', ascending=True)

# 3. สร้างกราฟแท่งแนวนอนเปรียบเทียบ
fig, ax = plt.subplots(figsize=(12, 6))

# กำหนดสี: สีส้มแดงสำหรับกลุ่มต่ำสุด และสีเขียวเข้มสำหรับกลุ่มสูงสุด
colors = ['#e6550d'] * 5 + ['#2ca02c'] * 5
bars = ax.barh(combined_df['intersection'], combined_df['total_vehicles'], color=colors, height=0.65)

# 4. ตกแต่งหัวข้อและป้ายกำกับ
ax.set_title('5 อันดับทางแยกที่มีปริมาณการจราจรสูงสุด (เขียว) และต่ำที่สุด (ส้ม)', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('ปริมาณรถรวมเฉลี่ย (คัน)', fontsize=12)
ax.set_ylabel('ทางแยก', fontsize=12)

# แสดงตัวเลขกำกับที่ปลายแท่งกราฟ
for bar in bars:
    width = bar.get_width()
    ax.text(width + 500, bar.get_y() + bar.get_height() / 2,
            f'{width:,.0f} คัน',
            va='center', ha='left', fontsize=10, fontweight='bold', color='#333333')

# ปรับแต่งตารางและกรอบ
ax.set_xlim(0, max(combined_df['total_vehicles']) * 1.18)
ax.grid(True, axis='x', linestyle='--', alpha=0.5)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# โค้ดสร้าง Dual Stacked Bar Chart (COVID Eras & Lockdown Status)
# ==============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# 1. โหลดข้อมูลและคำนวณสัดส่วน
file_path = 'bangkok_traffic_2560_Present (NotFinal).xlsx'
df = pd.read_excel(file_path)

vehicle_cols = ['passenger_car', 'pickup_van', 'large_bus', 'small_bus', 'truck', 'tuk_tuk']

# คำนวณช่วง COVID
covid_mix = df.groupby('covid_period')[vehicle_cols].sum()
covid_mix_pct = covid_mix.div(covid_mix.sum(axis=1), axis=0) * 100
period_mapping = {'ก่อนโควิด': 'Before', 'โควิด': 'During', 'หลังโควิด': 'After'}
covid_mix_pct['period_en'] = covid_mix_pct.index.map(period_mapping)
covid_mix_pct = covid_mix_pct.set_index('period_en').reindex(['Before', 'During', 'After'])

# คำนวณช่วง Lockdown vs Not Lockdown
lockdown_mix = df.groupby('Lockdown')[vehicle_cols].sum()
lockdown_mix_pct = lockdown_mix.div(lockdown_mix.sum(axis=1), axis=0) * 100
lockdown_mix_pct.index = ['Not Lockdown', 'Lockdown'] # 0: Not Lockdown, 1: Lockdown

# 2. ตั้งค่า Figure แบบ 1 แถว 2 คอลัมน์
fig = plt.figure(figsize=(16, 8.5), dpi=300)
gs = gridspec.GridSpec(1, 2, width_ratios=[1, 1], wspace=0.22)

colors = ['#1f77b4', '#2ca02c', '#9467bd', '#e377c2', '#bcbd22', '#17becf']
labels = ['Passenger car', 'Pickup / van', 'Large bus', 'Small bus', 'Truck', 'Tuk-tuk']

# --- Subplot A: COVID Eras ---
ax1 = fig.add_subplot(gs[0])
bottom1 = np.zeros(len(covid_mix_pct))
for i, col in enumerate(vehicle_cols):
    values = covid_mix_pct[col].values
    ax1.bar(covid_mix_pct.index, values, bottom=bottom1, label=labels[i], color=colors[i], width=0.42, edgecolor='white', linewidth=0.8)
    for j, val in enumerate(values):
        if val > 6:
            ax1.text(j, bottom1[j] + val / 2, f'{val:.1f}%', ha='center', va='center', color='white', fontweight='bold', fontsize=9.5)
    bottom1 += values

ax1.set_ylim(0, 108)
ax1.set_ylabel('Mean Vehicle Share (%)', fontsize=11, fontweight='bold')
ax1.set_title('(A) Vehicle Mix Share Across COVID Eras', fontsize=12, fontweight='bold', pad=12)
ax1.grid(True, axis='y', linestyle='--', alpha=0.3)
ax1.legend(loc='upper left', fontsize=8.5, frameon=True, facecolor='white', edgecolor='#e2e8f0')

# --- Subplot B: Lockdown vs Not Lockdown ---
ax2 = fig.add_subplot(gs[1])
bottom2 = np.zeros(len(lockdown_mix_pct))
for i, col in enumerate(vehicle_cols):
    values = lockdown_mix_pct[col].values
    ax2.bar(lockdown_mix_pct.index, values, bottom=bottom2, label=labels[i], color=colors[i], width=0.42, edgecolor='white', linewidth=0.8)
    for j, val in enumerate(values):
        if val > 6:
            ax2.text(j, bottom2[j] + val / 2, f'{val:.1f}%', ha='center', va='center', color='white', fontweight='bold', fontsize=9.5)
    bottom2 += values

ax2.set_ylim(0, 108)
ax2.set_ylabel('Mean Vehicle Share (%)', fontsize=11, fontweight='bold')
ax2.set_title('(B) Vehicle Mix Share: Lockdown vs Not Lockdown', fontsize=12, fontweight='bold', pad=12)
ax2.grid(True, axis='y', linestyle='--', alpha=0.3)

# 3. กล่องข้อความสรุปรายละเอียดค่าน้อยด้านล่าง
summary_text = (
    "Minor Shares Breakdown (%):\n"
    "• COVID Eras (Before / During / After) — Truck: 1.86% | 1.89% | 2.30%  |  Large Bus: 1.76% | 1.79% | 1.46%  |  Tuk-tuk: 1.64% | 1.37% | 1.17%  |  Small Bus: 0.24% | 0.25% | 0.12%\n"
    "• Lockdown Status — Not Lockdown: Truck 2.10%, Bus(L) 1.62%, Tuk 1.35%, Bus(S) 0.18%  |  Lockdown: Bus(L) 1.73%, Tuk 1.56%, Truck 1.54%, Bus(S) 0.21%"
)

fig.text(0.5, 0.04, summary_text, ha='center', fontsize=8.5,
         bbox=dict(boxstyle='round,pad=0.6', facecolor='#f8f9fa', edgecolor='#cbd5e1', alpha=0.95))

plt.subplots_adjust(bottom=0.25, top=0.90, left=0.08, right=0.95)
plt.show()
```

```python
# Let's create a dual chart showing (A) Mean Vehicle Volume by COVID Era and (B) Mean Vehicle Volume by Lockdown Status
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

file_path = 'bangkok_traffic_2560_Present (NotFinal).xlsx'
df = pd.read_excel(file_path)

vehicle_cols = ['passenger_car', 'pickup_van', 'large_bus', 'small_bus', 'truck', 'tuk_tuk']
colors = ['#1f77b4', '#2ca02c', '#9467bd', '#e377c2', '#bcbd22', '#17becf']
labels = ['Passenger car', 'Pickup / van', 'Large bus', 'Small bus', 'Truck', 'Tuk-tuk']

# 1. Calculate Mean Volume by COVID Era
covid_mean = df.groupby('covid_period')[vehicle_cols].mean()
period_mapping = {'ก่อนโควิด': 'Before', 'โควิด': 'During', 'หลังโควิด': 'After'}
covid_mean['period_en'] = covid_mean.index.map(period_mapping)
covid_mean = covid_mean.set_index('period_en').reindex(['Before', 'During', 'After'])

# 2. Calculate Mean Volume by Lockdown
lockdown_mean = df.groupby('Lockdown')[vehicle_cols].mean()
lockdown_mean.index = ['Not Lockdown', 'Lockdown']

fig = plt.figure(figsize=(16, 7.5), dpi=300)
gs = gridspec.GridSpec(1, 2, width_ratios=[1, 1], wspace=0.22)

# --- Subplot A: COVID Eras Volume ---
ax1 = fig.add_subplot(gs[0])
bottom1 = np.zeros(len(covid_mean))
for i, col in enumerate(vehicle_cols):
    values = covid_mean[col].values
    ax1.bar(covid_mean.index, values, bottom=bottom1, label=labels[i], color=colors[i], width=0.42, edgecolor='white', linewidth=0.8)
    for j, val in enumerate(values):
        if val > 300: # Only label if segment is large enough
            ax1.text(j, bottom1[j] + val / 2, f'{val:.0f}', ha='center', va='center', color='white', fontweight='bold', fontsize=9)
    bottom1 += values

ax1.set_ylabel('Mean Vehicle Volume (Count)', fontsize=11, fontweight='bold')
ax1.set_title('(A) Mean Vehicle Volume Across COVID Eras', fontsize=12, fontweight='bold', pad=12)
ax1.grid(True, axis='y', linestyle='--', alpha=0.3)
ax1.legend(loc='upper right', fontsize=8.5, frameon=True, facecolor='white', edgecolor='#e2e8f0')

# --- Subplot B: Lockdown Volume ---
ax2 = fig.add_subplot(gs[1])
bottom2 = np.zeros(len(lockdown_mean))
for i, col in enumerate(vehicle_cols):
    values = lockdown_mean[col].values
    ax2.bar(lockdown_mean.index, values, bottom=bottom2, label=labels[i], color=colors[i], width=0.42, edgecolor='white', linewidth=0.8)
    for j, val in enumerate(values):
        if val > 300:
            ax2.text(j, bottom2[j] + val / 2, f'{val:.0f}', ha='center', va='center', color='white', fontweight='bold', fontsize=9)
    bottom2 += values

ax2.set_ylabel('Mean Vehicle Volume (Count)', fontsize=11, fontweight='bold')
ax2.set_title('(B) Mean Vehicle Volume: Lockdown vs Not Lockdown', fontsize=12, fontweight='bold', pad=12)
ax2.grid(True, axis='y', linestyle='--', alpha=0.3)

# Add summary total table/box below
summary_text = (
    "Total Mean Volume Summary (Count per record):\n"
    f"• COVID Eras — Before: {covid_mean.sum(axis=1)['Before']:.1f}  |  During: {covid_mean.sum(axis=1)['During']:.1f}  |  After: {covid_mean.sum(axis=1)['After']:.1f}\n"
    f"• Lockdown Status — Not Lockdown: {lockdown_mean.sum(axis=1)['Not Lockdown']:.1f}  |  Lockdown: {lockdown_mean.sum(axis=1)['Lockdown']:.1f}"
)

fig.text(0.5, 0.04, summary_text, ha='center', fontsize=9, bbox=dict(boxstyle='round,pad=0.6', facecolor='#f8f9fa', edgecolor='#cbd5e1', alpha=0.95))

plt.subplots_adjust(bottom=0.22, top=0.90, left=0.08, right=0.95)
plt.savefig('volume_comparison_dual.png', dpi=300)
print("Saved volume comparison chart successfully!")
```

```python
import matplotlib.pyplot as plt
from matplotlib.offsetbox import AnnotationBbox, OffsetImage
import numpy as np
import pandas as pd
import urllib.request

# 1. โหลดข้อมูลจากไฟล์ Excel
file_path = "bangkok_traffic_2560_Present (NotFinal).xlsx"
df = pd.read_excel(file_path, sheet_name="traffic_data")

# 2. คำนวณปริมาณรถรวมทุกประเภท
vehicle_cols = [
    "passenger_car",
    "pickup_van",
    "large_bus",
    "small_bus",
    "truck",
    "tuk_tuk",
]
df["total_volume"] = df[vehicle_cols].sum(axis=1)

# แปลงชื่อช่วงเวลา COVID เป็นภาษาอังกฤษ
df["period_en"] = df["covid_period"].map(
    {
        "ก่อนโควิด": "Before COVID",
        "โควิด": "During COVID",
        "หลังโควิด": "After COVID",
    }
)

# คำนวณค่าเฉลี่ย Total Volume แยกตามช่วงเวลา
avg_volume = (
    df.groupby("period_en")["total_volume"]
    .mean()
    .reindex(["Before COVID", "During COVID", "After COVID"])
)

# 3. ดาวน์โหลดรูปไอคอนรถยนต์น่ารักๆ (ใส่ User-Agent เพื่อป้องกัน HTTP 403 Forbidden)
cute_car_url = "https://cdn-icons-png.flaticon.com/512/744/744465.png"
req = urllib.request.Request(
    cute_car_url, headers={"User-Agent": "Mozilla/5.0"}
)

with urllib.request.urlopen(req) as response, open(
    "car_icon_cute.png", "wb"
) as f:
    f.write(response.read())

# 4. พล็อต Pictogram
fig, ax = plt.subplots(figsize=(10, 6), dpi=300)

categories = avg_volume.index
values = avg_volume.values
car_img = plt.imread("car_icon_cute.png")

scale_factor = 500  # ไอคอน 1 รูป แทนรถประมาณ 500 คัน
x_positions = np.arange(len(categories))

for i, val in enumerate(values):
    num_icons = int(val / scale_factor)
    for j in range(num_icons):
        # ปรับ zoom เพื่อขยาย/ย่อรูปไอคอนรถให้น่ารักพอดี
        imagebox = OffsetImage(car_img, zoom=0.07)
        ab = AnnotationBbox(
            imagebox,
            (x_positions[i], j * 0.9),
            frameon=False,
            box_alignment=(0.5, 0.5),
        )
        ax.add_artist(ab)

    # แสดงตัวเลขค่าเฉลี่ยไว้ด้านบนสุด
    ax.text(
        x_positions[i],
        (num_icons * 0.9) + 0.5,
        f"{val:.1f}",
        ha="center",
        va="bottom",
        fontsize=11,
        fontweight="bold",
        color="#2c3e50",
    )

# ตกแต่งกราฟ
ax.set_xticks(x_positions)
ax.set_xticklabels(categories, fontsize=11, fontweight="bold")
ax.set_xlim(-0.5, len(categories) - 0.5)
ax.set_ylim(0, (max(values) / scale_factor * 0.9) + 2)

ax.set_title(
    "Traffic Volume Pictogram by COVID Period (1 Car Icon = 500 Vehicles)",
    fontsize=12,
    fontweight="bold",
    pad=20,
)
ax.set_ylabel(
    "Equivalent Traffic Volume Scale", fontsize=11, fontweight="bold"
)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.get_yaxis().set_ticks([])  # ซ่อนตัวเลขแกน Y เพื่อความสะอาดตา

plt.tight_layout()
plt.savefig("traffic_pictogram_cute.png", dpi=300)
print("ดาวน์โหลดรูปภาพและสร้างกราฟ Pictogram สำเร็จแล้ว!")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# 1. โหลดข้อมูล
file_path = 'bangkok_traffic_2560_Present (NotFinal).xlsx'
df = pd.read_excel(file_path, sheet_name='traffic_data')

# 2. เตรียมข้อมูล
vehicle_cols = ['passenger_car', 'pickup_van', 'large_bus', 'small_bus', 'truck', 'tuk_tuk']
veh_map = {'passenger_car': 'Passenger Car', 'pickup_van': 'Pickup/Van', 'truck': 'Truck',
           'large_bus': 'Large Bus', 'tuk_tuk': 'Tuk-Tuk', 'small_bus': 'Small Bus'}

df['period_en'] = df['covid_period'].map({'ก่อนโควิด': 'Before COVID', 'โควิด': 'During COVID', 'หลังโควิด': 'After COVID'})

melted = df.melt(id_vars=['period_en'], value_vars=vehicle_cols, var_name='vehicle', value_name='vol')
melted['vehicle'] = melted['vehicle'].map(veh_map)
avg_veh = melted.groupby(['vehicle', 'period_en'])['vol'].mean().unstack()

# 3. คำนวณแกน X และ Y
action_df = pd.DataFrame({
    'Current_Volume': avg_veh['After COVID'],
    'Rebound_Rate': (avg_veh['After COVID'] / avg_veh['Before COVID']) * 100
}).reset_index()

# 4. พล็อตกราฟ Action Priority Matrix
sns.set_theme(style="whitegrid")
plt.rcParams['font.sans-serif'] = ['DejaVu Sans', 'Arial', 'Helvetica']

fig, ax = plt.subplots(figsize=(14, 9), dpi=300)

x_threshold = action_df['Current_Volume'].median()
y_threshold = 100.0

# ระบายสีพื้นหลังแบ่ง Quadrant
ax.axvspan(x_threshold, action_df['Current_Volume'].max() * 1.2, ymin=0.5, ymax=1, color='#ffeaea', alpha=0.4)
ax.axvspan(0, x_threshold, ymin=0.5, ymax=1, color='#fff9e6', alpha=0.4)
ax.axvspan(0, x_threshold, ymin=0, ymax=0.5, color='#eafbf1', alpha=0.4)
ax.axvspan(x_threshold, action_df['Current_Volume'].max() * 1.2, ymin=0, ymax=0.5, color='#ebf5fb', alpha=0.4)

# วาดเส้นแบ่ง Quadrant
ax.axvline(x=x_threshold, color='#7f8c8d', linestyle='--', linewidth=1.5)
ax.axhline(y=y_threshold, color='#7f8c8d', linestyle='--', linewidth=1.5)

# พล็อตจุด Bubble
colors = sns.color_palette("husl", len(action_df))
scatter = ax.scatter(action_df['Current_Volume'], action_df['Rebound_Rate'],
                     s=action_df['Current_Volume'] * 0.15 + 400,
                     c=colors, alpha=0.8, edgecolor='white', linewidth=2)

# ใส่ชื่อประเภทรถ
for i, row in action_df.iterrows():
    ax.annotate(row['vehicle'],
                xy=(row['Current_Volume'], row['Rebound_Rate']),
                xytext=(0, 15),
                textcoords="offset points",
                fontsize=12, fontweight='bold', ha='center', color='#2c3e50')

# ==========================================
# 🌟 แก้ไข: ใช้ข้อความภาษาอังกฤษมาตรฐาน ไร้ปัญหากล่องสี่เหลี่ยม
# ==========================================
props = dict(boxstyle='round,pad=0.8', facecolor='white', alpha=0.9, edgecolor='none')

ax.text(0.97, 0.95, 'URGENT CONTROL\n(Congestion Fees / Route Limits)',
        transform=ax.transAxes, fontsize=12, fontweight='bold', color='#c0392b', ha='right', va='top', bbox=props)
ax.text(0.03, 0.95, 'MONITOR & REGULATE\n(Zoning / Time Restrictions)',
        transform=ax.transAxes, fontsize=12, fontweight='bold', color='#d35400', ha='left', va='top', bbox=props)
ax.text(0.03, 0.05, 'SUPPORT & REVITALIZE\n(Subsidies / System Upgrades)',
        transform=ax.transAxes, fontsize=12, fontweight='bold', color='#27ae60', ha='left', va='bottom', bbox=props)
ax.text(0.97, 0.05, 'MANAGE EFFICIENCY\n(Bottleneck Fixes / Smart AI Signals)',
        transform=ax.transAxes, fontsize=12, fontweight='bold', color='#2980b9', ha='right', va='bottom', bbox=props)
# ==========================================

# ตกแต่งกราฟ
ax.set_title("Strategic Traffic Action Priority Matrix\n(Based on Post-COVID Recovery Behavior)",
             fontsize=18, fontweight='bold', color='#1a365d', pad=20)
ax.set_xlabel("Current Traffic Volume (Average Count After COVID)", fontsize=12, fontweight='bold', labelpad=10)
ax.set_ylabel("Rebound Rate vs Before COVID (%)", fontsize=12, fontweight='bold', labelpad=10)

ax.set_xlim(-100, action_df['Current_Volume'].max() * 1.15)
ax.set_ylim(min(action_df['Rebound_Rate']) - 10, max(action_df['Rebound_Rate']) + 15)

sns.despine()
plt.tight_layout()
plt.savefig('traffic_action_plan_matrix_clean.png', dpi=300)
print("สร้างกราฟ Action Plan Matrix ฉบับภาษาอังกฤษสมบูรณ์ ไร้ปัญหากล่องสี่เหลี่ยมสำเร็จแล้ว!")
plt.show()
```
