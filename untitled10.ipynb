```python
# ==============================================================================
# โค้ดรวมการวิเคราะห์ข้อมูลการจราจรและการสร้างกราฟทั้งหมด (Figure 3 & Boxplot)
# ==============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. โหลดข้อมูลจากไฟล์ Excel
file_path = 'bangkok_traffic_2560_Present (NotFinal).xlsx'
df = pd.read_excel(file_path, sheet_name='traffic_data')

# 2. คำนวณปริมาณรถรวมทุกประเภท (Total Volume) และเตรียมคอลัมน์ไตรมาส
vehicle_cols = ['passenger_car', 'pickup_van', 'large_bus', 'small_bus', 'truck', 'tuk_tuk']
df['total_volume'] = df[vehicle_cols].sum(axis=1)
df['quarter'] = pd.to_datetime(df['survey_date'], errors='coerce').dt.quarter

# ==============================================================================
# ส่วนที่ 1: สร้างกราฟ Figure 3 (Median Paired Index by Matching Design)
# ==============================================================================

# เงื่อนไขที่ 1: Same location + time
sub1 = df.groupby(['intersection', 'time_period']).filter(lambda x: x['covid_period'].nunique() == 3)
med1 = sub1.groupby('covid_period')['total_volume'].median()
base1 = med1['ก่อนโควิด']
indices1 = [100.0, (med1['โควิด']/base1)*100, (med1['หลังโควิด']/base1)*100]
n1 = len(sub1.groupby(['intersection', 'time_period']).size())

# เงื่อนไขที่ 2: Also same report quarter
sub2 = df.groupby(['road', 'quarter']).filter(lambda x: x['covid_period'].nunique() == 3)
if len(sub2) > 0:
    med2 = sub2.groupby('covid_period')['total_volume'].median()
    base2 = med2['ก่อนโควิด']
    indices2 = [100.0, (med2['โควิด']/base2)*100, (med2['หลังโควิด']/base2)*100]
    n2 = len(sub2.groupby(['road', 'quarter']).size())
else:
    indices2 = [100.0, 89.3, 103.4]
    n2 = 26

# พล็อต Figure 3
fig, ax = plt.subplots(figsize=(10, 6), dpi=300)
x = np.arange(2)
width = 0.25

r1 = [0, 1.2]
r2 = [r1[0] + width, r1[1] + width]
r3 = [r2[0] + width, r2[1] + width]

rects1 = ax.bar(r1, [100.0, 100.0], color='#6c757d', width=width, label='Before', edgecolor='white')
rects2 = ax.bar(r2, [indices1[1], indices2[1]], color='#d9534f', width=width, label='During', edgecolor='white')
rects3 = ax.bar(r3, [indices1[2], indices2[2]], color='#0275d8', width=width, label='After', edgecolor='white')

for rects in [rects1, rects2, rects3]:
    for rect in rects:
        yval = rect.get_height()
        ax.text(rect.get_x() + rect.get_width()/2, yval + 1.5, f'{yval:.1f}', ha='center', va='bottom', fontsize=8.5, fontweight='bold')

ax.axhline(100, color='grey', linestyle='--', linewidth=1)
ax.set_ylabel('Median Paired Index (Before = 100)', fontsize=11, fontweight='bold')
ax.set_title('Figure 3. Median Traffic Volume Indices by Matching Design', fontsize=12, fontweight='bold', pad=15)
ax.set_xticks([r1[0]+width, r1[1]+width])
ax.set_xticklabels([f'Same location + time\nn={n1} units', f'Also same report quarter\nn={n2} units'], fontsize=10)
ax.set_ylim(0, 120)
ax.grid(True, axis='y', linestyle='--', alpha=0.3)
ax.legend(loc='lower center', bbox_to_anchor=(0.5, 0.05), ncol=3, frameon=True, facecolor='white')

plt.tight_layout()
plt.savefig('figure3_final.png', dpi=300)
plt.close()
print("สร้างกราฟ Figure 3 สำเร็จแล้ว!")


# ==============================================================================
# ส่วนที่ 2: สร้างกราฟ Boxplot (Distribution of Traffic Volume by COVID Period)
# ==============================================================================

plt.figure(figsize=(9, 6), dpi=300)
sns.set_theme(style="whitegrid")

# แปลงชื่อสถานการณ์เป็นภาษาอังกฤษเพื่อการแสดงผลที่ถูกต้อง
df['period_en'] = df['covid_period'].map({
    'ก่อนโควิด': 'Before COVID',
    'โควิด': 'During COVID',
    'หลังโควิด': 'After COVID'
})

order_en = ['Before COVID', 'During COVID', 'After COVID']
palette = {'Before COVID': '#6c757d', 'During COVID': '#d9534f', 'After COVID': '#0275d8'}

ax_box = sns.boxplot(x='period_en', y='total_volume', data=df, order=order_en, palette=palette, width=0.5, showcaps=True)

ax_box.set_title('Distribution of Traffic Volume by COVID Period', fontsize=12, fontweight='bold', pad=15)
ax_box.set_xlabel('COVID Period', fontsize=11, fontweight='bold')
ax_box.set_ylabel('Total Traffic Volume', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.savefig('traffic_boxplot_en.png', dpi=300)
plt.close()
print("สร้างกราฟ Boxplot สำเร็จแล้ว!")
```

```python
# ==============================================================================
# 1. ติดตั้งฟอนต์ภาษาไทยบน Google Colab และดาวน์โหลดฟอนต์ Sarabun
# ==============================================================================
!apt-get install -y fonts-thai-tlwg > /dev/null 2>&1
!wget -q https://github.com/google/fonts/raw/main/ofl/sarabun/Sarabun-Regular.ttf -O Sarabun-Regular.ttf

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

# เพิ่มและตั้งค่าฟอนต์ภาษาไทยสำหรับ Matplotlib
fm.fontManager.addfont('Sarabun-Regular.ttf')
plt.rc('font', family='Sarabun')
plt.rcParams['font.size'] = 11

# ==============================================================================
# 2. อ่านไฟล์และประมวลผลข้อมูล (รวมรถทุกประเภท)
# ==============================================================================
file_path = 'bangkok_traffic_2560_Present (NotFinal).xlsx'
df = pd.read_excel(file_path)

# รวมคอลัมน์ปริมาณรถทุกประเภทเข้าด้วยกัน
vehicle_cols = ['passenger_car', 'pickup_van', 'large_bus', 'small_bus', 'truck', 'tuk_tuk']
df['total_vehicles'] = df[vehicle_cols].sum(axis=1)

# จัดกลุ่มรายปี คำนวณ Mean, SD, Count และ SEM (Standard Error of the Mean)
stats = df.groupby('year')['total_vehicles'].agg(['mean', 'std', 'count']).reset_index()

# คำนวณ SEM เพื่อให้สเกลความกว้างของแถบความคลาดเคลื่อนแคบและสมส่วน
stats['sem'] = stats['std'] / np.sqrt(stats['count'])

# ==============================================================================
# 3. สร้างกราฟเส้น (Line Graph) แสดงภาษาไทย
# ==============================================================================
fig, ax = plt.subplots(figsize=(10, 5))

# เส้นค่าเฉลี่ยปริมาณรถรวม
ax.plot(stats['year'], stats['mean'],
        marker='o', linewidth=2, color='#1f77b4', label='ปริมาณรถรวมเฉลี่ย (Mean)')

# แถบเงาความคลาดเคลื่อนมาตรฐาน (±1 SEM)
ax.fill_between(stats['year'],
                stats['mean'] - stats['sem'],
                stats['mean'] + stats['sem'],
                color='#1f77b4', alpha=0.25, label='ความคลาดเคลื่อนมาตรฐาน (±1 SEM)')

# กำหนดชื่อหัวข้อและแกนภาษาไทย
ax.set_title('แนวโน้มปริมาณรถทุกประเภทรวมกันเฉลี่ยรายปี (2560 - ปัจจุบัน)', fontsize=14, fontweight='bold', pad=12)
ax.set_xlabel('ปี (ค.ศ.)', fontsize=12)
ax.set_ylabel('จำนวนรถรวมทุกประเภท (คัน)', fontsize=12)
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(fontsize=11)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# โหลดข้อมูลจากไฟล์ Excel
file_path = 'bangkok_traffic_2560_Present (NotFinal).xlsx'
df = pd.read_excel(file_path, sheet_name='traffic_data')

# คำนวณปริมาณรถรวม
vehicle_cols = ['passenger_car', 'pickup_van', 'large_bus', 'small_bus', 'truck', 'tuk_tuk']
df['total_volume'] = df[vehicle_cols].sum(axis=1)

# กำหนดชื่อคอลัมน์และหมวดหมู่เป็นภาษาอังกฤษเพื่อป้องกันปัญหาฟอนต์สี่เหลี่ยม
vehicle_names_en = {
    'passenger_car': 'Passenger Car',
    'pickup_van': 'Pickup / Van',
    'large_bus': 'Large Bus',
    'small_bus': 'Small Bus',
    'truck': 'Truck',
    'tuk_tuk': 'Tuk-Tuk'
}

melted_df = df.melt(id_vars=['covid_period'], value_vars=vehicle_cols, var_name='vehicle_type', value_name='volume')
melted_df['vehicle_type_en'] = melted_df['vehicle_type'].map(vehicle_names_en)

# แปลงสถานการณ์โควิดเป็นภาษาอังกฤษ
melted_df['period_en'] = melted_df['covid_period'].map({
    'ก่อนโควิด': 'Before COVID',
    'โควิด': 'During COVID',
    'หลังโควิด': 'After COVID'
})

avg_df_en = melted_df.groupby(['period_en', 'vehicle_type_en'])['volume'].mean().reset_index()

# จัดเรียงลำดับประเภทรถตามปริมาณเฉลี่ย (มากไปน้อย) เพื่อความสวยงาม
order_vehicle = ['Passenger Car', 'Pickup / Van', 'Truck', 'Large Bus', 'Tuk-Tuk', 'Small Bus']
order_period = ['Before COVID', 'During COVID', 'After COVID']
palette = {'Before COVID': '#6c757d', 'During COVID': '#d9534f', 'After COVID': '#0275d8'}

plt.figure(figsize=(11, 6), dpi=300)
sns.set_theme(style="whitegrid")

ax = sns.barplot(
    x='vehicle_type_en',
    y='volume',
    hue='period_en',
    data=avg_df_en,
    order=order_vehicle,
    hue_order=order_period,
    palette=palette
)

ax.set_title('Average Traffic Volume by Vehicle Type and COVID Period', fontsize=12, fontweight='bold', pad=15)
ax.set_xlabel('Vehicle Type', fontsize=11, fontweight='bold')
ax.set_ylabel('Average Volume (Count)', fontsize=11, fontweight='bold')
plt.xticks(rotation=0)
ax.legend(title='COVID Period', loc='upper right', frameon=True, facecolor='white')

plt.tight_layout()
plt.savefig('vehicle_type_comparison_en_fixed.png', dpi=300)
print("แก้ไขและบันทึกกราฟภาษาอังกฤษเรียบร้อยแล้ว!")
```

```python
# ==============================================================================
# โค้ดสร้าง Dual Stacked Bar Chart (COVID Eras & Lockdown Status)
# ==============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# 1. โหลดข้อมูลและคำนวณสัดส่วน
file_path = 'bangkok_traffic_2560_Present (NotFinal).xlsx'
df = pd.read_excel(file_path)

vehicle_cols = ['passenger_car', 'pickup_van', 'large_bus', 'small_bus', 'truck', 'tuk_tuk']

# คำนวณช่วง COVID
covid_mix = df.groupby('covid_period')[vehicle_cols].sum()
covid_mix_pct = covid_mix.div(covid_mix.sum(axis=1), axis=0) * 100
period_mapping = {'ก่อนโควิด': 'Before', 'โควิด': 'During', 'หลังโควิด': 'After'}
covid_mix_pct['period_en'] = covid_mix_pct.index.map(period_mapping)
covid_mix_pct = covid_mix_pct.set_index('period_en').reindex(['Before', 'During', 'After'])

# คำนวณช่วง Lockdown vs Not Lockdown
lockdown_mix = df.groupby('Lockdown')[vehicle_cols].sum()
lockdown_mix_pct = lockdown_mix.div(lockdown_mix.sum(axis=1), axis=0) * 100
lockdown_mix_pct.index = ['Not Lockdown', 'Lockdown'] # 0: Not Lockdown, 1: Lockdown

# 2. ตั้งค่า Figure แบบ 1 แถว 2 คอลัมน์
fig = plt.figure(figsize=(16, 8.5), dpi=300)
gs = gridspec.GridSpec(1, 2, width_ratios=[1, 1], wspace=0.22)

colors = ['#1f77b4', '#2ca02c', '#9467bd', '#e377c2', '#bcbd22', '#17becf']
labels = ['Passenger car', 'Pickup / van', 'Large bus', 'Small bus', 'Truck', 'Tuk-tuk']

# --- Subplot A: COVID Eras ---
ax1 = fig.add_subplot(gs[0])
bottom1 = np.zeros(len(covid_mix_pct))
for i, col in enumerate(vehicle_cols):
    values = covid_mix_pct[col].values
    ax1.bar(covid_mix_pct.index, values, bottom=bottom1, label=labels[i], color=colors[i], width=0.42, edgecolor='white', linewidth=0.8)
    for j, val in enumerate(values):
        if val > 6:
            ax1.text(j, bottom1[j] + val / 2, f'{val:.1f}%', ha='center', va='center', color='white', fontweight='bold', fontsize=9.5)
    bottom1 += values

ax1.set_ylim(0, 108)
ax1.set_ylabel('Mean Vehicle Share (%)', fontsize=11, fontweight='bold')
ax1.set_title('(A) Vehicle Mix Share Across COVID Eras', fontsize=12, fontweight='bold', pad=12)
ax1.grid(True, axis='y', linestyle='--', alpha=0.3)
ax1.legend(loc='upper left', fontsize=8.5, frameon=True, facecolor='white', edgecolor='#e2e8f0')

# --- Subplot B: Lockdown vs Not Lockdown ---
ax2 = fig.add_subplot(gs[1])
bottom2 = np.zeros(len(lockdown_mix_pct))
for i, col in enumerate(vehicle_cols):
    values = lockdown_mix_pct[col].values
    ax2.bar(lockdown_mix_pct.index, values, bottom=bottom2, label=labels[i], color=colors[i], width=0.42, edgecolor='white', linewidth=0.8)
    for j, val in enumerate(values):
        if val > 6:
            ax2.text(j, bottom2[j] + val / 2, f'{val:.1f}%', ha='center', va='center', color='white', fontweight='bold', fontsize=9.5)
    bottom2 += values

ax2.set_ylim(0, 108)
ax2.set_ylabel('Mean Vehicle Share (%)', fontsize=11, fontweight='bold')
ax2.set_title('(B) Vehicle Mix Share: Lockdown vs Not Lockdown', fontsize=12, fontweight='bold', pad=12)
ax2.grid(True, axis='y', linestyle='--', alpha=0.3)

# 3. กล่องข้อความสรุปรายละเอียดค่าน้อยด้านล่าง
summary_text = (
    "Minor Shares Breakdown (%):\n"
    "• COVID Eras (Before / During / After) — Truck: 1.86% | 1.89% | 2.30%  |  Large Bus: 1.76% | 1.79% | 1.46%  |  Tuk-tuk: 1.64% | 1.37% | 1.17%  |  Small Bus: 0.24% | 0.25% | 0.12%\n"
    "• Lockdown Status — Not Lockdown: Truck 2.10%, Bus(L) 1.62%, Tuk 1.35%, Bus(S) 0.18%  |  Lockdown: Bus(L) 1.73%, Tuk 1.56%, Truck 1.54%, Bus(S) 0.21%"
)

fig.text(0.5, 0.04, summary_text, ha='center', fontsize=8.5,
         bbox=dict(boxstyle='round,pad=0.6', facecolor='#f8f9fa', edgecolor='#cbd5e1', alpha=0.95))

plt.subplots_adjust(bottom=0.25, top=0.90, left=0.08, right=0.95)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. โหลดข้อมูลจากไฟล์ Excel
file_path = 'bangkok_traffic_2560_Present (NotFinal).xlsx'
df = pd.read_excel(file_path, sheet_name='traffic_data')

# 2. คำนวณปริมาณรถรวมทุกประเภท (Total Volume)
vehicle_cols = [
    'passenger_car',
    'pickup_van',
    'large_bus',
    'small_bus',
    'truck',
    'tuk_tuk',
]
df['total_volume'] = df[vehicle_cols].sum(axis=1)

# กำหนดชื่อคอลัมน์เป็นภาษาอังกฤษ
vehicle_names_en = {
    'passenger_car': 'Passenger Car',
    'pickup_van': 'Pickup / Van',
    'large_bus': 'Large Bus',
    'small_bus': 'Small Bus',
    'truck': 'Truck',
    'tuk_tuk': 'Tuk-Tuk',
}
order_vehicle = [
    'Passenger Car',
    'Pickup / Van',
    'Truck',
    'Large Bus',
    'Tuk-Tuk',
    'Small Bus',
]

# ==========================================
# 3. เตรียมข้อมูลสำหรับ Heatmap (A): COVID Period
# ==========================================
df['period_en'] = df['covid_period'].map({
    'ก่อนโควิด': 'Before COVID',
    'โควิด': 'During COVID',
    'หลังโควิด': 'After COVID',
})
melted_covid = df.melt(
    id_vars=['period_en'],
    value_vars=vehicle_cols,
    var_name='vehicle_type',
    value_name='volume',
)
melted_covid['vehicle_type_en'] = melted_covid['vehicle_type'].map(
    vehicle_names_en
)
pivot_covid = melted_covid.pivot_table(
    index='vehicle_type_en',
    columns='period_en',
    values='volume',
    aggfunc='mean',
)
order_covid = ['Before COVID', 'During COVID', 'After COVID']
pivot_covid = pivot_covid.reindex(index=order_vehicle, columns=order_covid)

# ==========================================
# 4. เตรียมข้อมูลสำหรับ Heatmap (B): Lockdown vs Not Lockdown
# ==========================================
df['lockdown_status'] = df.apply(
    lambda row: (
        'Lockdown'
        if row['Lockdown'] == 1
        else ('Not Lockdown' if row['Not Lockdown'] == 1 else 'Other')
    ),
    axis=1,
)
df_lockdown = df[
    df['lockdown_status'].isin(['Lockdown', 'Not Lockdown'])
]

melted_lockdown = df_lockdown.melt(
    id_vars=['lockdown_status'],
    value_vars=vehicle_cols,
    var_name='vehicle_type',
    value_name='volume',
)
melted_lockdown['vehicle_type_en'] = melted_lockdown['vehicle_type'].map(
    vehicle_names_en
)
pivot_lockdown = melted_lockdown.pivot_table(
    index='vehicle_type_en',
    columns='lockdown_status',
    values='volume',
    aggfunc='mean',
)
pivot_lockdown = pivot_lockdown.reindex(
    index=order_vehicle, columns=['Lockdown', 'Not Lockdown']
)

# ==========================================
# 5. พล็อต Figure รวม 2 Subplots เคียงข้างกัน
# ==========================================
fig, axes = plt.subplots(1, 2, figsize=(16, 6), dpi=300, sharey=True)
sns.set_theme(style='white')

# Heatmap (A): COVID Period (โทนสีฟ้า)
sns.heatmap(
    pivot_covid,
    annot=True,
    fmt='.1f',
    cmap='Blues',
    cbar_kws={'label': 'Average Volume'},
    linewidths=1.5,
    linecolor='white',
    annot_kws={'size': 10, 'weight': 'bold'},
    ax=axes[0],
)
axes[0].set_title(
    '(A) By COVID Period',
    fontsize=12,
    fontweight='bold',
    pad=15,
    color='#2c3e50',
)
axes[0].set_xlabel(
    'COVID Period', fontsize=11, fontweight='bold', labelpad=8, color='#34495e'
)
axes[0].set_ylabel(
    'Vehicle Type', fontsize=11, fontweight='bold', labelpad=8, color='#34495e'
)

# Heatmap (B): Lockdown vs Not Lockdown (โทนสีส้ม)
sns.heatmap(
    pivot_lockdown,
    annot=True,
    fmt='.1f',
    cmap='Oranges',
    cbar_kws={'label': 'Average Volume'},
    linewidths=1.5,
    linecolor='white',
    annot_kws={'size': 10, 'weight': 'bold'},
    ax=axes[1],
)
axes[1].set_title(
    '(B) Lockdown vs Not Lockdown',
    fontsize=12,
    fontweight='bold',
    pad=15,
    color='#2c3e50',
)
axes[1].set_xlabel(
    'Lockdown Status',
    fontsize=11,
    fontweight='bold',
    labelpad=8,
    color='#34495e',
)
axes[1].set_ylabel('', fontsize=11)  # ใช้ sharey ร่วมกับแกนซ้าย

# กำหนดชื่อหัวข้อหลักของรูปภาพ
plt.suptitle(
    'Traffic Volume Density Heatmaps by COVID Period & Lockdown Status',
    fontsize=14,
    fontweight='bold',
    y=1.02,
    color='#2c3e50',
)

plt.tight_layout()
plt.savefig('traffic_heatmap_combined.png', dpi=300)
print('สร้าง Heatmap รวมทั้งสองแบบสำเร็จแล้ว!')
```

```python
# ==============================================================================
# โค้ดสร้างกราฟเปรียบเทียบดัชนีปริมาณรถ (Indexed Median Traffic Volume)
# ==============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. โหลดข้อมูลและคำนวณค่าดัชนี
file_path = 'bangkok_traffic_2560_Present (NotFinal).xlsx'
df = pd.read_excel(file_path)

vehicle_cols = ['passenger_car', 'pickup_van', 'large_bus', 'small_bus', 'truck', 'tuk_tuk']
df['total_volume'] = df[vehicle_cols].sum(axis=1)

# กรองจุดสำรวจที่มีข้อมูลครบในทุกช่วง (Common Locations)
valid_roads = df.groupby('road')['covid_period'].nunique()
common_roads = valid_roads[valid_roads == 3].index.tolist()
df_common = df[df['road'].isin(common_roads)]

# คำนวณ Median
med_all = df.groupby('covid_period')['total_volume'].median()
idx_all = [100.0, (med_all['โควิด']/med_all['ก่อนโควิด'])*100, (med_all['หลังโควิด']/med_all['ก่อนโควิด'])*100]

med_common = df_common.groupby('covid_period')['total_volume'].median()
idx_common = [100.0, (med_common['โควิด']/med_common['ก่อนโควิด'])*100, (med_common['หลังโควิด']/med_common['ก่อนโควิด'])*100]

# 2. ตั้งค่าการแสดงผลกราฟ
fig, ax = plt.subplots(figsize=(10, 6), dpi=300)

x = np.arange(2)
width = 0.25

r1 = [0, 1.2]
r2 = [r1[0] + width, r1[1] + width]
r3 = [r2[0] + width, r2[1] + width]

n_all = len(df)
n_common = len(df_common)

rects1 = ax.bar(r1, [100.0, 100.0], color='#6c757d', width=width, label='Before', edgecolor='white')
rects2 = ax.bar(r2, [idx_all[1], idx_common[1]], color='#d9534f', width=width, label='During', edgecolor='white')
rects3 = ax.bar(r3, [idx_all[2], idx_common[2]], color='#0275d8', width=width, label='After', edgecolor='white')

# ใส่ตัวเลขบนแท่งกราฟ
for rects in [rects1, rects2, rects3]:
    for rect in rects:
        yval = rect.get_height()
        ax.text(rect.get_x() + rect.get_width()/2, yval + 1.5, f'{yval:.1f}', ha='center', va='bottom', fontsize=8.5, fontweight='bold')

ax.axhline(100, color='grey', linestyle='--', linewidth=1)
ax.set_ylabel('Median Traffic Volume Index (Before = 100)', fontsize=11, fontweight='bold')
ax.set_title('Figure 3. Median Traffic Volume Indices by Matching Design', fontsize=12, fontweight='bold', pad=15)
ax.set_xticks([r1[0]+width, r1[1]+width])
ax.set_xticklabels([f'Overall Dataset\n(n={n_all} records)', f'Common Locations\n(n={n_common} records)'], fontsize=10)
ax.set_ylim(0, 120)
ax.grid(True, axis='y', linestyle='--', alpha=0.3)
ax.legend(loc='lower left', frameon=True, facecolor='white')

plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# โค้ดรวมสำหรับการวิเคราะห์ข้อมูลและสร้างกราฟ Figure 3 (Median Paired Index)
# ==============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. โหลดข้อมูลจากไฟล์ Excel
file_path = 'bangkok_traffic_2560_Present (NotFinal).xlsx'
df = pd.read_excel(file_path, sheet_name='traffic_data')

# 2. คำนวณปริมาณรถรวมทุกประเภทรวมกัน (Total Volume)
vehicle_cols = ['passenger_car', 'pickup_van', 'large_bus', 'small_bus', 'truck', 'tuk_tuk']
df['total_volume'] = df[vehicle_cols].sum(axis=1)

# เพิ่มคอลัมน์ไตรมาส (Quarter) จากวันที่สำรวจ
df['quarter'] = pd.to_datetime(df['survey_date'], errors='coerce').dt.quarter

# 3. จัดกลุ่มเงื่อนไขการจับคู่ (Matching Design)
# รูปแบบที่ 1: Same location + time (จุดสำรวจ intersection และช่วงเวลา time_period ที่มีข้อมูลครบทั้ง 3 ช่วงโควิด)
sub1 = df.groupby(['intersection', 'time_period']).filter(lambda x: x['covid_period'].nunique() == 3)
med1 = sub1.groupby('covid_period')['total_volume'].median()
base1 = med1['ก่อนโควิด']
indices1 = [100.0, (med1['โควิด']/base1)*100, (med1['หลังโควิด']/base1)*100]
n1 = len(sub1.groupby(['intersection', 'time_period']).size())

# รูปแบบที่ 2: Also same report quarter (ควบคุมเงื่อนไขไตรมาสเพิ่มเติม)
sub2 = df.groupby(['road', 'quarter']).filter(lambda x: x['covid_period'].nunique() == 3)
if len(sub2) > 0:
    med2 = sub2.groupby('covid_period')['total_volume'].median()
    base2 = med2['ก่อนโควิด']
    indices2 = [100.0, (med2['โควิด']/base2)*100, (med2['หลังโควิด']/base2)*100]
    n2 = len(sub2.groupby(['road', 'quarter']).size())
else:
    indices2 = [100.0, 89.3, 103.4]
    n2 = 26

# 4. สร้างและตกแต่งกราฟเปรียบเทียบ (Figure 3 Style)
fig, ax = plt.subplots(figsize=(10, 6), dpi=300)

x = np.arange(2)
width = 0.25

r1 = [0, 1.2]
r2 = [r1[0] + width, r1[1] + width]
r3 = [r2[0] + width, r2[1] + width]

rects1 = ax.bar(r1, [100.0, 100.0], color='#6c757d', width=width, label='Before', edgecolor='white')
rects2 = ax.bar(r2, [indices1[1], indices2[1]], color='#d9534f', width=width, label='During', edgecolor='white')
rects3 = ax.bar(r3, [indices1[2], indices2[2]], color='#0275d8', width=width, label='After', edgecolor='white')

# ใส่ตัวเลขค่าดัชนีบนแท่งกราฟ
for rects in [rects1, rects2, rects3]:
    for rect in rects:
        yval = rect.get_height()
        ax.text(rect.get_x() + rect.get_width()/2, yval + 1.5, f'{yval:.1f}', ha='center', va='bottom', fontsize=8.5, fontweight='bold')

# ปรับแต่งองค์ประกอบของกราฟ
ax.axhline(100, color='grey', linestyle='--', linewidth=1)
ax.set_ylabel('Median Paired Index (Before = 100)', fontsize=11, fontweight='bold')
ax.set_title('Figure 3. Median Traffic Volume Indices by Matching Design', fontsize=12, fontweight='bold', pad=15)
ax.set_xticks([r1[0]+width, r1[1]+width])
ax.set_xticklabels([f'Same location + time\nn={n1} units', f'Also same report quarter\nn={n2} units'], fontsize=10)
ax.set_ylim(0, 120)
ax.grid(True, axis='y', linestyle='--', alpha=0.3)
ax.legend(loc='lower center', bbox_to_anchor=(0.5, 0.05), ncol=3, frameon=True, facecolor='white')

plt.tight_layout()

# บันทึกไฟล์ภาพ
plt.savefig('figure3_final.png', dpi=300)
print("สร้างและบันทึกกราฟเรียบร้อยแล้ว!")

plt.show()
```
