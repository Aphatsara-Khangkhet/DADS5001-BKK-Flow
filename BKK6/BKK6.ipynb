Notebook นี้ฝังโค้ดและฟอนต์ไว้แล้ว ไม่ต้องคัดลอก src เอง เปิดใน Colab แล้ว Run all และเลือกไฟล์ Excel ชีต traffic_data ของคุณ ผลสร้างในโฟลเดอร์ใหม่เสมอ อ่าน README ที่สร้างและทวน audit ก่อนนำเสนอ

หากเป็นไฟล์เดิมที่ SHA256 ตรงกัน จะสร้างรายงานจากผลคำนวณที่ตรวจแล้ว หากเปลี่ยนข้อมูล จะตรวจและคำนวณใหม่ทั้งหมด ผลต้นทางยังไม่ใช่ข้อมูลที่รับรองทุกแถว และข้อจำกัดเรื่องสิทธิใช้ข้อมูล/นิยามล็อกดาวน์ยังคงอยู่

```python
import os, sys, subprocess
os.environ["OPENBLAS_NUM_THREADS"] = "1"
if "google.colab" in sys.modules:
    subprocess.check_call([sys.executable, "-m", "pip", "-q", "install", "pandas", "numpy", "matplotlib", "openpyxl"])
```

```python
from pathlib import Path
import base64, io, zipfile, datetime, hashlib, json
RUNTIME = Path.cwd() / ("bkk6_" + datetime.datetime.now().strftime("%Y%m%d_%H%M%S_%f"))
RUNTIME.mkdir()
PAYLOAD = ''
with zipfile.ZipFile(io.BytesIO(base64.b64decode(PAYLOAD))) as z:
    for item in z.infolist():
        assert (RUNTIME / item.filename).resolve().is_relative_to(RUNTIME.resolve())
    z.extractall(RUNTIME)
sys.path.insert(0, str(RUNTIME / "src"))
import readme_project
print("โค้ดพร้อมใช้งาน:", RUNTIME)
```

```python
INPUT = None  # หากรันในเครื่อง เปลี่ยนเป็น Path(r"C:/.../new.xlsx")
if INPUT is None:
    try:
        from google.colab import files
    except ImportError:
        raise ValueError("ระบุ INPUT เป็นพาธ Excel ของคุณในเซลล์นี้ก่อน")
    uploaded = files.upload()
    choices = [name for name in uploaded if name.lower().endswith(".xlsx")]
    if len(choices) != 1:
        raise ValueError("กรุณาเลือกไฟล์ .xlsx เพียงหนึ่งไฟล์")
    INPUT = Path(choices[0]).resolve()
INPUT = Path(INPUT).resolve()
assert INPUT.is_file(), INPUT
print("ไฟล์:", INPUT.name)
```

```python
sha = hashlib.sha256(INPUT.read_bytes()).hexdigest()
cached = json.loads((RUNTIME / "cache/presentation_data.json").read_text(encoding="utf-8"))
reuse = RUNTIME / "cache" if cached["stats"]["source_sha256"] == sha else None
print("ใช้ผลตรวจเดิมที่ตรง SHA256" if reuse else "ไฟล์เปลี่ยน: ตรวจข้อมูลและคำนวณใหม่")
REPORT = readme_project.build(INPUT, reviewed_report=reuse)
print("README:", REPORT / "README.md")
```

```python
from IPython.display import display, Markdown, Image, FileLink
display(FileLink(str(REPORT / "README.md")))
for png in sorted((REPORT / "fig").glob("*.png")):
    display(Image(filename=str(png)))
```
